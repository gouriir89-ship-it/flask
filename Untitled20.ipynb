# Diabetes Prediction: Analysis, Modeling, and Deployment
This notebook demonstrates a complete machine learning pipeline using the `diabetes_prediction_dataset.csv`. We will explore the data, preprocess it, train and compare models, and set up a Flask application code for deployment.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = '/content/diabetes_prediction_dataset.csv'
df = pd.read_csv(file_path)

# Display basic information
print("Dataset Shape:", df.shape)
display(df.head())
print(df.info())
print(df.describe())
```

```python
# Visualizing the distribution of the target variable and missing values
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='diabetes')
plt.title('Distribution of Diabetes Target Variable')
plt.show()

# Check for null values
print("Missing Values:")
print(df.isnull().sum())
```
